```r
# Código R CORREGIDO para análisis de producción de petróleo y gas en Google Colab
# Basado en la estructura real de los datos

library(readxl)
library(dplyr)
library(ggplot2)
library(lubridate)
library(scales)
library(gridExtra)
library(knitr)
library(DT)
library(tidyr)

# ================================
# CARGAR Y LIMPIAR DATOS
# ================================

print("=== CARGANDO Y PREPARANDO DATOS ===")

# Leer datos de petróleo
oil_data_raw <- read_excel("Datos_Tratados.xlsx", sheet = "Producción Oil m3", skip = 1)

# Leer datos de gas
gas_data_raw <- read_excel("Datos_Tratados.xlsx", sheet = "Producción Gas miles m3", skip = 1)

# Función para limpiar y reestructurar datos
clean_and_reshape_data <- function(data, data_type = "oil") {

  # Identificar las columnas correctas basándonos en la estructura observada
  if (data_type == "oil") {
    # Para petróleo: empresa, cuenca, provincia, yacimiento, concesion, on-off shore, concepto, año, meses...
    colnames(data) <- c("empresa", "cuenca", "provincia", "yacimiento", "concesion",
                       "on_off_shore", "concepto", "año", "Enero", "Febrero", "Marzo",
                       "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre",
                       "Octubre", "Noviembre", "Diciembre", "total_general")
  } else {
    # Para gas: empresa, año, yacimiento, concesion, provincia, on-off shore, cuenca, concepto, meses...
    colnames(data) <- c("empresa", "año", "yacimiento", "concesion", "provincia",
                       "on_off_shore", "cuenca", "concepto", "Enero", "Febrero", "Marzo",
                       "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre",
                       "Octubre", "Noviembre", "Diciembre", "total_general")
  }

  # Filtrar solo producción primaria (no condensado ni otros conceptos)
  if (data_type == "oil") {
    data <- data %>%
      filter(grepl("Producción primaria|Producción Oil", concepto, ignore.case = TRUE) |
             is.na(concepto))
  } else {
    data <- data %>%
      filter(grepl("Gas|Producción", concepto, ignore.case = TRUE) |
             is.na(concepto))
  }

  # Convertir año a numérico
  data$año <- as.numeric(data$año)

  # Convertir de formato ancho a largo
  data_long <- data %>%
    pivot_longer(cols = c("Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
                         "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"),
                names_to = "mes_nombre",
                values_to = "produccion") %>%
    filter(!is.na(produccion), produccion != "") %>%
    mutate(
      produccion = as.numeric(produccion),
      mes_numero = match(mes_nombre, month.name[c(1:12)]),
      mes_numero = case_when(
        mes_nombre == "Enero" ~ 1,
        mes_nombre == "Febrero" ~ 2,
        mes_nombre == "Marzo" ~ 3,
        mes_nombre == "Abril" ~ 4,
        mes_nombre == "Mayo" ~ 5,
        mes_nombre == "Junio" ~ 6,
        mes_nombre == "Julio" ~ 7,
        mes_nombre == "Agosto" ~ 8,
        mes_nombre == "Septiembre" ~ 9,
        mes_nombre == "Octubre" ~ 10,
        mes_nombre == "Noviembre" ~ 11,
        mes_nombre == "Diciembre" ~ 12,
        TRUE ~ NA_real_
      ),
      fecha = as.Date(paste(año, mes_numero, "01", sep = "-")),
      año_mes = floor_date(fecha, "month")
    ) %>%
    filter(!is.na(produccion), produccion > 0, !is.na(año), año >= 2018, año <= 2024)

  return(data_long)
}

# Limpiar datos
print("Limpiando datos de petróleo...")
oil_data_clean <- clean_and_reshape_data(oil_data_raw, "oil")

print("Limpiando datos de gas...")
gas_data_clean <- clean_and_reshape_data(gas_data_raw, "gas")

print(paste("Registros de petróleo procesados:", nrow(oil_data_clean)))
print(paste("Registros de gas procesados:", nrow(gas_data_clean)))

# Verificar empresas disponibles
print("=== EMPRESAS DISPONIBLES ===")
print("Empresas en datos de petróleo que contienen 'PAMPA':")
pampa_empresas <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE)) %>%
  distinct(empresa) %>%
  pull(empresa)
print(pampa_empresas)

print("Empresas en datos de gas que contienen 'COMBUSTIBLES':")
cgc_empresas <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE)) %>%
  distinct(empresa) %>%
  pull(empresa)
print(cgc_empresas)

# ================================
# CONSIGNAS DE PETRÓLEO (OIL)
# ================================

# Consigna inicial: Evolución mensual de producción de oil PAMPA ENERGIA S.A. (2021-2023)
print("=== Consigna: Evolución mensual PAMPA ENERGIA S.A. (2021-2023) ===")

pampa_oil_2021_2023 <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE),
         año >= 2021, año <= 2023) %>%
  group_by(año_mes) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(año_mes)

# Mostrar tabla
if (nrow(pampa_oil_2021_2023) > 0) {
  print(kable(pampa_oil_2021_2023, caption = "Producción mensual de oil - PAMPA ENERGIA S.A. (2021-2023)"))

  # Gráfico
  p1 <- ggplot(pampa_oil_2021_2023, aes(x = año_mes, y = produccion_total)) +
    geom_line(color = "blue", size = 1) +
    geom_point(color = "red", size = 2) +
    labs(title = "Evolución Mensual - Producción Oil PAMPA ENERGIA S.A. (2021-2023)",
         x = "Fecha", y = "Producción Oil (m³)") +
    theme_minimal() +
    scale_x_date(date_labels = "%Y-%m", date_breaks = "3 months") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
    scale_y_continuous(labels = comma_format())

  print(p1)
} else {
  print("No se encontraron datos para PAMPA ENERGIA en el período 2021-2023")
}

# Consigna: Distribución por cuencas PAMPA ENERGIA S.A. (2020-2023)
print("=== Distribución por cuencas PAMPA ENERGIA S.A. (2020-2023) ===")

pampa_cuencas_2020_2023 <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE),
         año >= 2020, año <= 2023) %>%
  group_by(cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total))

if (nrow(pampa_cuencas_2020_2023) > 0) {
  print(kable(pampa_cuencas_2020_2023, caption = "Producción total por cuenca - PAMPA ENERGIA S.A. (2020-2023)"))

  # Gráfico
  p2 <- ggplot(pampa_cuencas_2020_2023, aes(x = reorder(cuenca, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "steelblue") +
    coord_flip() +
    labs(title = "Distribución Producción Oil por Cuenca - PAMPA ENERGIA S.A. (2020-2023)",
         x = "Cuenca", y = "Producción Total Oil (m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format())

  print(p2)
}

# Consigna: Distribución por yacimientos PAMPA ENERGIA S.A. (2022-2023)
print("=== Distribución por yacimientos PAMPA ENERGIA S.A. (2022-2023) ===")

pampa_yacimientos_2022_2023 <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE),
         año >= 2022, año <= 2023) %>%
  group_by(yacimiento) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total))

if (nrow(pampa_yacimientos_2022_2023) > 0) {
  print(kable(head(pampa_yacimientos_2022_2023, 10), caption = "Top 10 Yacimientos - PAMPA ENERGIA S.A. (2022-2023)"))

  # Gráfico para los primeros 15 yacimientos
  top_yacimientos <- head(pampa_yacimientos_2022_2023, 15)

  p3 <- ggplot(top_yacimientos, aes(x = reorder(yacimiento, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "darkgreen") +
    coord_flip() +
    labs(title = "Top 15 Yacimientos - PAMPA ENERGIA S.A. (2022-2023)",
         x = "Yacimiento", y = "Producción Total Oil (m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format()) +
    theme(axis.text.y = element_text(size = 8))

  print(p3)
}

# Consigna 1: Evolución por cuencas PAMPA ENERGIA S.A. (2018-2023)
print("=== Consigna 1: Evolución por cuencas PAMPA ENERGIA S.A. (2018-2023) ===")

pampa_cuencas_evolucion <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE),
         año >= 2018, año <= 2023) %>%
  group_by(año_mes, cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop')

if (nrow(pampa_cuencas_evolucion) > 0) {
  # Mostrar tabla resumida
  pampa_cuencas_resumen <- pampa_cuencas_evolucion %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion_total, na.rm = TRUE),
              promedio_mensual = mean(produccion_total, na.rm = TRUE),
              .groups = 'drop') %>%
    arrange(desc(produccion_total))

  print(kable(pampa_cuencas_resumen, caption = "Resumen producción por cuenca - PAMPA ENERGIA S.A. (2018-2023)"))

  # Gráfico
  p4 <- ggplot(pampa_cuencas_evolucion, aes(x = año_mes, y = produccion_total, color = cuenca)) +
    geom_line(size = 1) +
    geom_point(size = 1.5) +
    labs(title = "Evolución Producción Oil por Cuenca - PAMPA ENERGIA S.A. (2018-2023)",
         x = "Fecha", y = "Producción Oil (m³)", color = "Cuenca") +
    theme_minimal() +
    scale_x_date(date_labels = "%Y", date_breaks = "1 year") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
    scale_y_continuous(labels = comma_format())

  print(p4)
}

# Consigna 2: Top 5 mediciones mensuales PAMPA ENERGIA S.A. (2018-2023)
print("=== Consigna 2: Top 5 mediciones mensuales PAMPA ENERGIA S.A. (2018-2023) ===")

pampa_top5_mensual <- oil_data_clean %>%
  filter(grepl("PAMPA", empresa, ignore.case = TRUE),
         año >= 2018, año <= 2023) %>%
  group_by(año_mes) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

if (nrow(pampa_top5_mensual) > 0) {
  print(kable(pampa_top5_mensual, caption = "Top 5 mediciones mensuales - PAMPA ENERGIA S.A. (2018-2023)"))

  # Gráfico
  p5 <- ggplot(pampa_top5_mensual, aes(x = reorder(format(año_mes, "%Y-%m"), produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "orange") +
    coord_flip() +
    labs(title = "Top 5 Mediciones Mensuales Oil - PAMPA ENERGIA S.A. (2018-2023)",
         x = "Fecha", y = "Producción Oil (m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format())

  print(p5)
}

# ================================
# CONSIGNAS DE GAS
# ================================

# Consigna 3: Evolución mensual gas COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023)
print("=== Consigna 3: Evolución mensual gas COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023) ===")

cgc_gas_2018_2023 <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
         año >= 2018, año <= 2023) %>%
  group_by(año_mes) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(año_mes)

if (nrow(cgc_gas_2018_2023) > 0) {
  print(kable(head(cgc_gas_2018_2023, 10), caption = "Producción mensual gas - COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023)"))

  # Gráfico
  p6 <- ggplot(cgc_gas_2018_2023, aes(x = año_mes, y = produccion_total)) +
    geom_line(color = "purple", size = 1) +
    geom_point(color = "red", size = 2) +
    labs(title = "Evolución Mensual - Producción Gas COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023)",
         x = "Fecha", y = "Producción Gas (miles m³)") +
    theme_minimal() +
    scale_x_date(date_labels = "%Y", date_breaks = "1 year") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
    scale_y_continuous(labels = comma_format())

  print(p6)
} else {
  print("No se encontraron datos para COMPAÑÍA GENERAL DE COMBUSTIBLES")
}

# Consigna 4: Distribución por cuencas gas CGC (2021-2022)
print("=== Consigna 4: Distribución por cuencas gas CGC (2021-2022) ===")

cgc_cuencas_2021_2022 <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
         año >= 2021, año <= 2022) %>%
  group_by(cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total))

if (nrow(cgc_cuencas_2021_2022) > 0) {
  print(kable(cgc_cuencas_2021_2022, caption = "Producción total gas por cuenca - CGC (2021-2022)"))

  # Gráfico
  p7 <- ggplot(cgc_cuencas_2021_2022, aes(x = reorder(cuenca, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "lightblue") +
    coord_flip() +
    labs(title = "Distribución Producción Gas por Cuenca - CGC (2021-2022)",
         x = "Cuenca", y = "Producción Total Gas (miles m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format())

  print(p7)
}

# Consigna 5: Distribución por yacimientos gas CGC (2019-2023)
print("=== Consigna 5: Distribución por yacimientos gas CGC (2019-2023) ===")

cgc_yacimientos_2019_2023 <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
         año >= 2019, año <= 2023) %>%
  group_by(yacimiento) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total))

if (nrow(cgc_yacimientos_2019_2023) > 0) {
  print(kable(head(cgc_yacimientos_2019_2023, 10), caption = "Top 10 Yacimientos gas - CGC (2019-2023)"))

  # Gráfico
  top_yac_gas <- head(cgc_yacimientos_2019_2023, 10)
  p8 <- ggplot(top_yac_gas, aes(x = reorder(yacimiento, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "darkcyan") +
    coord_flip() +
    labs(title = "Top 10 Yacimientos Gas - CGC (2019-2023)",
         x = "Yacimiento", y = "Producción Total Gas (miles m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format()) +
    theme(axis.text.y = element_text(size = 8))

  print(p8)
}

# Consigna 6: Evolución por cuencas gas CGC (2018-2020)
print("=== Consigna 6: Evolución por cuencas gas CGC (2018-2020) ===")

cgc_cuencas_evolucion_2018_2020 <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
         año >= 2018, año <= 2020) %>%
  group_by(año_mes, cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop')

if (nrow(cgc_cuencas_evolucion_2018_2020) > 0) {
  # Mostrar tabla resumida
  cgc_cuencas_resumen_2018_2020 <- cgc_cuencas_evolucion_2018_2020 %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion_total, na.rm = TRUE),
              promedio_mensual = mean(produccion_total, na.rm = TRUE),
              .groups = 'drop') %>%
    arrange(desc(produccion_total))

  print(kable(cgc_cuencas_resumen_2018_2020, caption = "Resumen producción gas por cuenca - CGC (2018-2020)"))

  # Gráfico
  p9 <- ggplot(cgc_cuencas_evolucion_2018_2020, aes(x = año_mes, y = produccion_total, color = cuenca)) +
    geom_line(size = 1) +
    geom_point(size = 1.5) +
    labs(title = "Evolución Producción Gas por Cuenca - CGC (2018-2020)",
         x = "Fecha", y = "Producción Gas (miles m³)", color = "Cuenca") +
    theme_minimal() +
    scale_x_date(date_labels = "%Y", date_breaks = "6 months") +
    theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
    scale_y_continuous(labels = comma_format())

  print(p9)
}

# Consigna 7: Top 5 mediciones mensuales gas CGC (2019-2023)
print("=== Consigna 7: Top 5 mediciones mensuales gas CGC (2019-2023) ===")

cgc_top5_mensual_gas <- gas_data_clean %>%
  filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
         año >= 2019, año <= 2023) %>%
  group_by(año_mes) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

if (nrow(cgc_top5_mensual_gas) > 0) {
  print(kable(cgc_top5_mensual_gas, caption = "Top 5 mediciones mensuales gas - CGC (2019-2023)"))

  # Gráfico
  p10 <- ggplot(cgc_top5_mensual_gas, aes(x = reorder(format(año_mes, "%Y-%m"), produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "darkred") +
    coord_flip() +
    labs(title = "Top 5 Mediciones Mensuales Gas - CGC (2019-2023)",
         x = "Fecha", y = "Producción Gas (miles m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format())

  print(p10)
}

# Consigna 8: Top 5 empresas oil Neuquén 2024
print("=== Consigna 8: Top 5 empresas oil Neuquén 2024 ===")

top5_empresas_oil_neuquen_2024 <- oil_data_clean %>%
  filter(grepl("NEUQUÉN|NEUQUEN", provincia, ignore.case = TRUE),
         año == 2024) %>%
  group_by(empresa) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

if (nrow(top5_empresas_oil_neuquen_2024) > 0) {
  print(kable(top5_empresas_oil_neuquen_2024, caption = "Top 5 empresas oil Neuquén 2024"))

  # Gráfico
  p11 <- ggplot(top5_empresas_oil_neuquen_2024, aes(x = reorder(empresa, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "gold") +
    coord_flip() +
    labs(title = "Top 5 Empresas Producción Oil - Neuquén 2024",
         x = "Empresa", y = "Producción Total Oil (m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format()) +
    theme(axis.text.y = element_text(size = 7))

  print(p11)
} else {
  print("No se encontraron datos para Neuquén en 2024")
}

# Consigna 9: Top 5 empresas gas Santa Cruz 2022
print("=== Consigna 9: Top 5 empresas gas Santa Cruz 2022 ===")

top5_empresas_gas_santacruz_2022 <- gas_data_clean %>%
  filter(grepl("SANTA CRUZ", provincia, ignore.case = TRUE),
         año == 2022) %>%
  group_by(empresa) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

if (nrow(top5_empresas_gas_santacruz_2022) > 0) {
  print(kable(top5_empresas_gas_santacruz_2022, caption = "Top 5 empresas gas Santa Cruz 2022"))

  # Gráfico
  p12 <- ggplot(top5_empresas_gas_santacruz_2022, aes(x = reorder(empresa, produccion_total), y = produccion_total)) +
    geom_bar(stat = "identity", fill = "lightgreen") +
    coord_flip() +
    labs(title = "Top 5 Empresas Producción Gas - Santa Cruz 2022",
         x = "Empresa", y = "Producción Total Gas (miles m³)") +
    theme_minimal() +
    scale_y_continuous(labels = comma_format()) +
    theme(axis.text.y = element_text(size = 7))

  print(p12)
} else {
  print("No se encontraron datos para Santa Cruz en 2022")
}

# ================================
# CONSIGNAS FINALES (SOLO TABLAS)
# ================================

# Consigna 10: Concesión con mayor producción oil 2024
print("=== Consigna 10: Concesión con mayor producción oil 2024 ===")

top_concesion_oil_2024 <- oil_data_clean %>%
  filter(año == 2024) %>%
  group_by(concesion) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(1)

if (nrow(top_concesion_oil_2024) > 0) {
  print(kable(top_concesion_oil_2024, caption = "Concesión con mayor producción oil 2024"))

  print(paste("La concesión con mayor producción de oil en 2024 fue:",
              top_concesion_oil_2024$concesion[1],
              "con un total de",
              format(top_concesion_oil_2024$produccion_total[1], big.mark = ","),
              "m³"))
} else {
  print("No se encontraron datos para 2024")
}

# Consigna 11: Concesión con mayor producción gas 2022
print("=== Consigna 11: Concesión con mayor producción gas 2022 ===")

top_concesion_gas_2022 <- gas_data_clean %>%
  filter(año == 2022) %>%
  group_by(concesion) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(1)

if (nrow(top_concesion_gas_2022) > 0) {
  print(kable(top_concesion_gas_2022, caption = "Concesión con mayor producción gas 2022"))

  print(paste("La concesión con mayor producción de gas en 2022 fue:",
              top_concesion_gas_2022$concesion[1],
              "con un total de",
              format(top_concesion_gas_2022$produccion_total[1], big.mark = ","),
              "miles de m³"))
} else {
  print("No se encontraron datos para 2022")
}

# Consigna 12: Identificar datos faltantes
print("=== Consigna 12: Identificar observaciones con datos faltantes ===")

print("ANÁLISIS DE DATOS FALTANTES - PETRÓLEO:")
oil_missing <- oil_data_clean %>%
  summarise_all(~sum(is.na(.))) %>%
  gather(columna, valores_faltantes) %>%
  filter(valores_faltantes > 0)

if (nrow(oil_missing) > 0) {
  print(kable(oil_missing, caption = "Resumen de valores faltantes en datos de petróleo"))

  # Mostrar observaciones con datos faltantes
  oil_with_na <- oil_data_clean[!complete.cases(oil_data_clean), ]

  if (nrow(oil_with_na) > 0) {
    print(paste("Número total de observaciones con datos faltantes en petróleo:", nrow(oil_with_na)))
    print(kable(head(oil_with_na, 20), caption = "Primeras 20 observaciones con datos faltantes - Petróleo"))
  }
} else {
  print("No se encontraron valores faltantes en los datos de petróleo.")
}

print("ANÁLISIS DE DATOS FALTANTES - GAS:")
gas_missing <- gas_data_clean %>%
  summarise_all(~sum(is.na(.))) %>%
  gather(columna, valores_faltantes) %>%
  filter(valores_faltantes > 0)

if (nrow(gas_missing) > 0) {
  print(kable(gas_missing, caption = "Resumen de valores faltantes en datos de gas"))

  # Mostrar observaciones con datos faltantes
  gas_with_na <- gas_data_clean[!complete.cases(gas_data_clean), ]

  if (nrow(gas_with_na) > 0) {
    print(paste("Número total de observaciones con datos faltantes en gas:", nrow(gas_with_na)))
    print(kable(head(gas_with_na, 20), caption = "Primeras 20 observaciones con datos faltantes - Gas"))
  }
} else {
  print("No se encontraron valores faltantes en los datos de gas.")
}

# Análisis detallado por columna
print("ANÁLISIS DETALLADO POR COLUMNA:")

analyze_column_missing <- function(data, column_name, data_type) {
  if (column_name %in% names(data)) {
    missing_count <- sum(is.na(data[[column_name]]))
    total_count <- nrow(data)
    percentage <- round((missing_count / total_count) * 100, 2)

    cat(sprintf("%s - %s: %d valores faltantes de %d total (%.2f%%)\n",
                data_type, column_name, missing_count, total_count, percentage))

    if (missing_count > 0) {
      missing_rows <- data[is.na(data[[column_name]]), ]
      return(head(missing_rows, 5))  # Solo mostrar primeras 5 filas
    }
  }
  return(NULL)
}

# Columnas a analizar
columns_to_check <- c("empresa", "cuenca", "provincia", "yacimiento", "concesion", "produccion", "año", "fecha")

cat("\nPETRÓLEO:\n")
for (col in columns_to_check) {
  oil_missing_col <- analyze_column_missing(oil_data_clean, col, "Petróleo")
}

cat("\nGAS:\n")
for (col in columns_to_check) {
  gas_missing_col <- analyze_column_missing(gas_data_clean, col, "Gas")
}

# Verificar registros con producción cero o valores extraños
print("\nANÁLISIS DE VALORES ATÍPICOS:")

# Petróleo - valores cero
oil_zeros <- oil_data_clean %>%
  filter(produccion == 0) %>%
  nrow()

# Petróleo - valores negativos
oil_negative <- oil_data_clean %>%
  filter(produccion < 0) %>%
  nrow()

# Gas - valores cero
gas_zeros <- gas_data_clean %>%
  filter(produccion == 0) %>%
  nrow()

# Gas - valores negativos
gas_negative <- gas_data_clean %>%
  filter(produccion < 0) %>%
  nrow()

cat(sprintf("Registros de petróleo con producción = 0: %d\n", oil_zeros))
cat(sprintf("Registros de petróleo con producción < 0: %d\n", oil_negative))
cat(sprintf("Registros de gas con producción = 0: %d\n", gas_zeros))
cat(sprintf("Registros de gas con producción < 0: %d\n", gas_negative))

# Mostrar ejemplos de registros problemáticos si existen
if (oil_negative > 0) {
  print("Ejemplos de registros de petróleo con valores negativos:")
  oil_neg_examples <- oil_data_clean %>%
    filter(produccion < 0) %>%
    head(5)
  print(kable(oil_neg_examples))
}

if (gas_negative > 0) {
  print("Ejemplos de registros de gas con valores negativos:")
  gas_neg_examples <- gas_data_clean %>%
    filter(produccion < 0) %>%
    head(5)
  print(kable(gas_neg_examples))
}

# ================================
# RESUMEN FINAL Y ESTADÍSTICAS
# ================================

print("=== RESUMEN FINAL DE ANÁLISIS ===")

# Estadísticas generales
cat("\nESTADÍSTICAS GENERALES:\n")
cat(sprintf("Total de observaciones de petróleo: %d\n", nrow(oil_data_clean)))
cat(sprintf("Total de observaciones de gas: %d\n", nrow(gas_data_clean)))

# Rangos de fechas
cat(sprintf("Rango de años en datos de petróleo: %d - %d\n",
            min(oil_data_clean$año, na.rm = TRUE),
            max(oil_data_clean$año, na.rm = TRUE)))

cat(sprintf("Rango de años en datos de gas: %d - %d\n",
            min(gas_data_clean$año, na.rm = TRUE),
            max(gas_data_clean$año, na.rm = TRUE)))

# Número de empresas únicas
cat(sprintf("Número de empresas únicas en petróleo: %d\n",
            length(unique(oil_data_clean$empresa))))

cat(sprintf("Número de empresas únicas en gas: %d\n",
            length(unique(gas_data_clean$empresa))))

# Número de cuencas únicas
cat(sprintf("Número de cuencas únicas en petróleo: %d\n",
            length(unique(oil_data_clean$cuenca))))

cat(sprintf("Número de cuencas únicas en gas: %d\n",
            length(unique(gas_data_clean$cuenca))))

# Estadísticas de producción
cat("\nESTADÍSTICAS DE PRODUCCIÓN:\n")
cat("Petróleo (m³):\n")
print(summary(oil_data_clean$produccion))

cat("\nGas (miles m³):\n")
print(summary(gas_data_clean$produccion))

# Top empresas productoras
print("\nTOP 10 EMPRESAS PRODUCTORAS DE PETRÓLEO (2018-2024):")
top_oil_companies <- oil_data_clean %>%
  group_by(empresa) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(10)

print(kable(top_oil_companies, caption = "Top 10 empresas productoras de petróleo"))

print("\nTOP 10 EMPRESAS PRODUCTORAS DE GAS (2018-2024):")
top_gas_companies <- gas_data_clean %>%
  group_by(empresa) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(10)

print(kable(top_gas_companies, caption = "Top 10 empresas productoras de gas"))

# Top cuencas productoras
print("\nTOP 5 CUENCAS PRODUCTORAS DE PETRÓLEO:")
top_oil_basins <- oil_data_clean %>%
  group_by(cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

print(kable(top_oil_basins, caption = "Top 5 cuencas productoras de petróleo"))

print("\nTOP 5 CUENCAS PRODUCTORAS DE GAS:")
top_gas_basins <- gas_data_clean %>%
  group_by(cuenca) %>%
  summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
  arrange(desc(produccion_total)) %>%
  head(5)

print(kable(top_gas_basins, caption = "Top 5 cuencas productoras de gas"))

# ================================
# FUNCIONES AUXILIARES
# ================================

# Función para mostrar todos los gráficos juntos (opcional)
show_all_plots <- function() {
  # Esta función puede ser llamada para mostrar todos los gráficos en una sola visualización
  # Útil para reportes

  plots_list <- list()

  # Agregar plots que existan
  if (exists("p1")) plots_list <- append(plots_list, list(p1))
  if (exists("p2")) plots_list <- append(plots_list, list(p2))
  if (exists("p3")) plots_list <- append(plots_list, list(p3))
  if (exists("p4")) plots_list <- append(plots_list, list(p4))
  if (exists("p5")) plots_list <- append(plots_list, list(p5))
  if (exists("p6")) plots_list <- append(plots_list, list(p6))
  if (exists("p7")) plots_list <- append(plots_list, list(p7))
  if (exists("p8")) plots_list <- append(plots_list, list(p8))
  if (exists("p9")) plots_list <- append(plots_list, list(p9))
  if (exists("p10")) plots_list <- append(plots_list, list(p10))
  if (exists("p11")) plots_list <- append(plots_list, list(p11))
  if (exists("p12")) plots_list <- append(plots_list, list(p12))

  # Mostrar plots en grupos de 4
  if (length(plots_list) > 0) {
    for (i in seq(1, length(plots_list), by = 4)) {
      end_idx <- min(i + 3, length(plots_list))
      current_plots <- plots_list[i:end_idx]

      if (length(current_plots) == 4) {
        grid.arrange(current_plots[[1]], current_plots[[2]],
                    current_plots[[3]], current_plots[[4]], ncol = 2, nrow = 2)
      } else if (length(current_plots) == 3) {
        grid.arrange(current_plots[[1]], current_plots[[2]],
                    current_plots[[3]], ncol = 2, nrow = 2)
      } else if (length(current_plots) == 2) {
        grid.arrange(current_plots[[1]], current_plots[[2]], ncol = 2)
      } else {
        print(current_plots[[1]])
      }
    }
  }
}

# Función para generar reporte en formato HTML
generate_html_report <- function() {
  # Esta función generaría un reporte HTML con todos los resultados
  cat("Para generar un reporte HTML completo, ejecute:\n")
  cat("rmarkdown::render('nombre_archivo.Rmd', output_file = 'reporte_petroleo_gas.html')\n")
}

# Función para exportar datos procesados
export_processed_data <- function() {
  # Exportar datos limpios a CSV
  write.csv(oil_data_clean, "datos_petroleo_procesados.csv", row.names = FALSE)
  write.csv(gas_data_clean, "datos_gas_procesados.csv", row.names = FALSE)

  cat("Datos procesados exportados a:\n")
  cat("- datos_petroleo_procesados.csv\n")
  cat("- datos_gas_procesados.csv\n")
}

print("=== ANÁLISIS COMPLETADO ===")

# ================================
# INSTRUCCIONES FINALES
# ================================

print("FUNCIONES DISPONIBLES:")
print("1. show_all_plots() - Muestra todos los gráficos generados")
print("2. export_processed_data() - Exporta los datos procesados a CSV")
print("3. generate_html_report() - Instrucciones para generar reporte HTML")

print("\nNOTAS IMPORTANTES:")
print("- Todos los gráficos y tablas han sido generados basándose en los datos disponibles")
print("- Si alguna consigna no muestra resultados, es porque no hay datos para ese período/empresa")
print("- Los datos han sido filtrados para incluir solo valores de producción > 0")
print("- Las fechas se construyeron combinando año y mes de cada registro")

print("\nPARA GUARDAR GRÁFICOS:")
print("Usa ggsave() para cada gráfico, por ejemplo:")
print("ggsave('grafico1.png', p1, width = 10, height = 6, dpi = 300)")

print("\n=== FIN DEL ANÁLISIS ===")
```

```r
# ================================
# PIPELINE COMPLETO - ANÁLISIS DE PRODUCCIÓN DE PETRÓLEO Y GAS
# ================================

# ================================
# PASO 1: CONFIGURACIÓN E INSTALACIÓN
# ================================

pipeline_step <- function(step_name, code_block) {
  cat("\n", rep("=", 60), "\n")
  cat("EJECUTANDO:", step_name, "\n")
  cat(rep("=", 60), "\n")

  result <- tryCatch({
    code_block()
    cat("✓ COMPLETADO:", step_name, "\n")
    TRUE
  }, error = function(e) {
    cat("✗ ERROR en", step_name, ":", e$message, "\n")
    FALSE
  })

  return(result)
}

# PASO 1: Configuración
pipeline_step("CONFIGURACIÓN E INSTALACIÓN DE LIBRERÍAS", {

  # Función para instalar y cargar paquetes
  install_and_load <- function(packages) {
    for (pkg in packages) {
      if (!require(pkg, character.only = TRUE, quietly = TRUE)) {
        install.packages(pkg, dependencies = TRUE, repos = "http://cran.rstudio.com/")
        library(pkg, character.only = TRUE)
      }
    }
  }

  # Lista de paquetes necesarios
  required_packages <- c("readxl", "dplyr", "ggplot2", "lubridate", "scales",
                        "gridExtra", "knitr", "DT", "tidyr", "stringr")

  cat("Instalando y cargando paquetes necesarios...\n")
  install_and_load(required_packages)

  cat("Librerías cargadas exitosamente\n")
})

# ================================
# PASO 2: CARGA Y LIMPIEZA DE DATOS
# ================================

pipeline_step("CARGA Y LIMPIEZA DE DATOS", {

  # Verificar que el archivo existe
  if (!file.exists("Datos_Tratados.xlsx")) {
    stop("Archivo 'Datos_Tratados.xlsx' no encontrado. Por favor súbelo al directorio de trabajo.")
  }

  cat("Leyendo datos de Excel...\n")

  # Leer datos de petróleo
  oil_data_raw <- read_excel("Datos_Tratados.xlsx", sheet = "Producción Oil m3", skip = 1)
  cat("✓ Datos de petróleo cargados:", nrow(oil_data_raw), "registros\n")

  # Leer datos de gas
  gas_data_raw <- read_excel("Datos_Tratados.xlsx", sheet = "Producción Gas miles m3", skip = 1)
  cat("✓ Datos de gas cargados:", nrow(gas_data_raw), "registros\n")

  # Función para limpiar y reestructurar datos
  clean_and_reshape_data <- function(data, data_type = "oil") {

    # Identificar las columnas correctas basándonos en la estructura observada
    if (data_type == "oil") {
      # Para petróleo: empresa, cuenca, provincia, yacimiento, concesion, on-off shore, concepto, año, meses...
      colnames(data) <- c("empresa", "cuenca", "provincia", "yacimiento", "concesion",
                         "on_off_shore", "concepto", "año", "Enero", "Febrero", "Marzo",
                         "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre",
                         "Octubre", "Noviembre", "Diciembre", "total_general")
    } else {
      # Para gas: empresa, año, yacimiento, concesion, provincia, on-off shore, cuenca, concepto, meses...
      colnames(data) <- c("empresa", "año", "yacimiento", "concesion", "provincia",
                         "on_off_shore", "cuenca", "concepto", "Enero", "Febrero", "Marzo",
                         "Abril", "Mayo", "Junio", "Julio", "Agosto", "Septiembre",
                         "Octubre", "Noviembre", "Diciembre", "total_general")
    }

    # Filtrar solo producción primaria (no condensado ni otros conceptos)
    if (data_type == "oil") {
      data <- data %>%
        filter(grepl("Producción primaria|Producción Oil|Petróleo", concepto, ignore.case = TRUE) |
               is.na(concepto) | concepto == "")
    } else {
      data <- data %>%
        filter(grepl("Gas|Producción", concepto, ignore.case = TRUE) |
               is.na(concepto) | concepto == "")
    }

    # Convertir año a numérico
    data$año <- as.numeric(data$año)

    # Convertir de formato ancho a largo
    data_long <- data %>%
      pivot_longer(cols = c("Enero", "Febrero", "Marzo", "Abril", "Mayo", "Junio",
                           "Julio", "Agosto", "Septiembre", "Octubre", "Noviembre", "Diciembre"),
                  names_to = "mes_nombre",
                  values_to = "produccion") %>%
      filter(!is.na(produccion), produccion != "", produccion != "0") %>%
      mutate(
        produccion = as.numeric(produccion),
        mes_numero = case_when(
          mes_nombre == "Enero" ~ 1,
          mes_nombre == "Febrero" ~ 2,
          mes_nombre == "Marzo" ~ 3,
          mes_nombre == "Abril" ~ 4,
          mes_nombre == "Mayo" ~ 5,
          mes_nombre == "Junio" ~ 6,
          mes_nombre == "Julio" ~ 7,
          mes_nombre == "Agosto" ~ 8,
          mes_nombre == "Septiembre" ~ 9,
          mes_nombre == "Octubre" ~ 10,
          mes_nombre == "Noviembre" ~ 11,
          mes_nombre == "Diciembre" ~ 12,
          TRUE ~ NA_real_
        ),
        fecha = as.Date(paste(año, mes_numero, "01", sep = "-")),
        año_mes = floor_date(fecha, "month")
      ) %>%
      filter(!is.na(produccion), produccion > 0, !is.na(año), año >= 2018, año <= 2024) %>%
      # Limpiar nombres de empresas
      mutate(
        empresa = str_trim(toupper(empresa)),
        cuenca = str_trim(toupper(cuenca)),
        provincia = str_trim(toupper(provincia)),
        yacimiento = str_trim(toupper(yacimiento)),
        concesion = str_trim(toupper(concesion))
      )

    return(data_long)
  }

  # Limpiar datos
  cat("Procesando datos de petróleo...\n")
  oil_data_clean <<- clean_and_reshape_data(oil_data_raw, "oil")
  cat("✓ Datos de petróleo procesados:", nrow(oil_data_clean), "registros\n")

  cat("Procesando datos de gas...\n")
  gas_data_clean <<- clean_and_reshape_data(gas_data_raw, "gas")
  cat("✓ Datos de gas procesados:", nrow(gas_data_clean), "registros\n")

  # Verificar empresas disponibles
  pampa_empresas <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE)) %>%
    distinct(empresa) %>%
    pull(empresa)

  cgc_empresas <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE)) %>%
    distinct(empresa) %>%
    pull(empresa)

  cat("Empresas PAMPA encontradas:", length(pampa_empresas), "\n")
  cat("Empresas CGC encontradas:", length(cgc_empresas), "\n")

})

# ================================
# PASO 3: ANÁLISIS DE PETRÓLEO - PAMPA ENERGIA
# ================================

pipeline_step("ANÁLISIS DE PETRÓLEO - PAMPA ENERGIA", {

  # Crear lista para almacenar todos los gráficos
  graficos <<- list()
  tablas <<- list()

  # Consigna inicial: Evolución mensual de producción de oil PAMPA ENERGIA S.A. (2021-2023)
  cat("Consigna inicial: Evolución mensual PAMPA ENERGIA (2021-2023)\n")

  pampa_oil_2021_2023 <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE),
           año >= 2021, año <= 2023) %>%
    group_by(año_mes) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(año_mes)

  if (nrow(pampa_oil_2021_2023) > 0) {
    tablas$pampa_mensual_2021_2023 <<- pampa_oil_2021_2023

    graficos$p1 <<- ggplot(pampa_oil_2021_2023, aes(x = año_mes, y = produccion_total)) +
      geom_line(color = "blue", size = 1) +
      geom_point(color = "red", size = 2) +
      labs(title = "Evolución Mensual - Producción Oil PAMPA ENERGIA S.A. (2021-2023)",
           x = "Fecha", y = "Producción Oil (m³)") +
      theme_minimal() +
      scale_x_date(date_labels = "%Y-%m", date_breaks = "3 months") +
      theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
      scale_y_continuous(labels = comma_format())

    cat("✓ Gráfico de evolución mensual PAMPA creado\n")
  }

  # Distribución por cuencas PAMPA ENERGIA S.A. (2020-2023)
  cat("Distribución por cuencas PAMPA ENERGIA (2020-2023)\n")

  pampa_cuencas_2020_2023 <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE),
           año >= 2020, año <= 2023) %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total))

  if (nrow(pampa_cuencas_2020_2023) > 0) {
    tablas$pampa_cuencas_2020_2023 <<- pampa_cuencas_2020_2023

    graficos$p2 <<- ggplot(pampa_cuencas_2020_2023, aes(x = reorder(cuenca, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "steelblue") +
      coord_flip() +
      labs(title = "Distribución Producción Oil por Cuenca - PAMPA ENERGIA S.A. (2020-2023)",
           x = "Cuenca", y = "Producción Total Oil (m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format())

    cat("✓ Gráfico de distribución por cuencas PAMPA creado\n")
  }

  # Distribución por yacimientos PAMPA ENERGIA S.A. (2022-2023)
  cat("Distribución por yacimientos PAMPA ENERGIA (2022-2023)\n")

  pampa_yacimientos_2022_2023 <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE),
           año >= 2022, año <= 2023) %>%
    group_by(yacimiento) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total))

  if (nrow(pampa_yacimientos_2022_2023) > 0) {
    tablas$pampa_yacimientos_2022_2023 <<- head(pampa_yacimientos_2022_2023, 15)

    graficos$p3 <<- ggplot(head(pampa_yacimientos_2022_2023, 15),
                          aes(x = reorder(yacimiento, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "darkgreen") +
      coord_flip() +
      labs(title = "Top 15 Yacimientos - PAMPA ENERGIA S.A. (2022-2023)",
           x = "Yacimiento", y = "Producción Total Oil (m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format()) +
      theme(axis.text.y = element_text(size = 8))

    cat("✓ Gráfico de distribución por yacimientos PAMPA creado\n")
  }

})

# ================================
# PASO 4: CONSIGNAS NUMERADAS DE PETRÓLEO
# ================================

pipeline_step("CONSIGNAS NUMERADAS DE PETRÓLEO", {

  # Consigna 1: Evolución por cuencas PAMPA ENERGIA S.A. (2018-2023)
  cat("Consigna 1: Evolución por cuencas PAMPA (2018-2023)\n")

  pampa_cuencas_evolucion <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE),
           año >= 2018, año <= 2023) %>%
    group_by(año_mes, cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop')

  if (nrow(pampa_cuencas_evolucion) > 0) {
    pampa_cuencas_resumen <- pampa_cuencas_evolucion %>%
      group_by(cuenca) %>%
      summarise(produccion_total = sum(produccion_total, na.rm = TRUE),
                promedio_mensual = mean(produccion_total, na.rm = TRUE),
                .groups = 'drop') %>%
      arrange(desc(produccion_total))

    tablas$consigna1_resumen <<- pampa_cuencas_resumen

    graficos$p4 <<- ggplot(pampa_cuencas_evolucion, aes(x = año_mes, y = produccion_total, color = cuenca)) +
      geom_line(size = 1) +
      geom_point(size = 1.5) +
      labs(title = "Evolución Producción Oil por Cuenca - PAMPA ENERGIA S.A. (2018-2023)",
           x = "Fecha", y = "Producción Oil (m³)", color = "Cuenca") +
      theme_minimal() +
      scale_x_date(date_labels = "%Y", date_breaks = "1 year") +
      theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
      scale_y_continuous(labels = comma_format())

    cat("✓ Consigna 1 completada\n")
  }

  # Consigna 2: Top 5 mediciones mensuales PAMPA ENERGIA S.A. (2018-2023)
  cat("Consigna 2: Top 5 mediciones mensuales PAMPA (2018-2023)\n")

  pampa_top5_mensual <- oil_data_clean %>%
    filter(grepl("PAMPA", empresa, ignore.case = TRUE),
           año >= 2018, año <= 2023) %>%
    group_by(año_mes) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  if (nrow(pampa_top5_mensual) > 0) {
    tablas$consigna2_top5 <<- pampa_top5_mensual

    graficos$p5 <<- ggplot(pampa_top5_mensual, aes(x = reorder(format(año_mes, "%Y-%m"), produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "orange") +
      coord_flip() +
      labs(title = "Top 5 Mediciones Mensuales Oil - PAMPA ENERGIA S.A. (2018-2023)",
           x = "Fecha", y = "Producción Oil (m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format())

    cat("✓ Consigna 2 completada\n")
  }

})

# ================================
# PASO 5: ANÁLISIS DE GAS - CGC
# ================================

pipeline_step("ANÁLISIS DE GAS - COMPAÑÍA GENERAL DE COMBUSTIBLES", {

  # Consigna 3: Evolución mensual gas COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023)
  cat("Consigna 3: Evolución mensual gas CGC (2018-2023)\n")

  cgc_gas_2018_2023 <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
           año >= 2018, año <= 2023) %>%
    group_by(año_mes) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(año_mes)

  if (nrow(cgc_gas_2018_2023) > 0) {
    tablas$consigna3_cgc_mensual <<- head(cgc_gas_2018_2023, 15)

    graficos$p6 <<- ggplot(cgc_gas_2018_2023, aes(x = año_mes, y = produccion_total)) +
      geom_line(color = "purple", size = 1) +
      geom_point(color = "red", size = 2) +
      labs(title = "Evolución Mensual - Producción Gas COMPAÑÍA GENERAL DE COMBUSTIBLES S.A. (2018-2023)",
           x = "Fecha", y = "Producción Gas (miles m³)") +
      theme_minimal() +
      scale_x_date(date_labels = "%Y", date_breaks = "1 year") +
      theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
      scale_y_continuous(labels = comma_format())

    cat("✓ Consigna 3 completada\n")
  } else {
    cat("⚠ No se encontraron datos para CGC\n")
  }

  # Consigna 4: Distribución por cuencas gas CGC (2021-2022)
  cat("Consigna 4: Distribución por cuencas gas CGC (2021-2022)\n")

  cgc_cuencas_2021_2022 <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
           año >= 2021, año <= 2022) %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total))

  if (nrow(cgc_cuencas_2021_2022) > 0) {
    tablas$consigna4_cgc_cuencas <<- cgc_cuencas_2021_2022

    graficos$p7 <<- ggplot(cgc_cuencas_2021_2022, aes(x = reorder(cuenca, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "lightblue") +
      coord_flip() +
      labs(title = "Distribución Producción Gas por Cuenca - CGC (2021-2022)",
           x = "Cuenca", y = "Producción Total Gas (miles m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format())

    cat("✓ Consigna 4 completada\n")
  }

  # Consignas 5, 6, 7 de gas
  cat("Completando consignas restantes de gas...\n")

  # Consigna 5: Yacimientos CGC (2019-2023)
  cgc_yacimientos_2019_2023 <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
           año >= 2019, año <= 2023) %>%
    group_by(yacimiento) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total))

  if (nrow(cgc_yacimientos_2019_2023) > 0) {
    tablas$consigna5_cgc_yacimientos <<- head(cgc_yacimientos_2019_2023, 10)

    graficos$p8 <<- ggplot(head(cgc_yacimientos_2019_2023, 10),
                          aes(x = reorder(yacimiento, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "darkcyan") +
      coord_flip() +
      labs(title = "Top 10 Yacimientos Gas - CGC (2019-2023)",
           x = "Yacimiento", y = "Producción Total Gas (miles m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format()) +
      theme(axis.text.y = element_text(size = 8))
  }

  # Consigna 6: Evolución cuencas CGC (2018-2020)
  cgc_cuencas_evolucion_2018_2020 <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
           año >= 2018, año <= 2020) %>%
    group_by(año_mes, cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop')

  if (nrow(cgc_cuencas_evolucion_2018_2020) > 0) {
    cgc_cuencas_resumen_2018_2020 <- cgc_cuencas_evolucion_2018_2020 %>%
      group_by(cuenca) %>%
      summarise(produccion_total = sum(produccion_total, na.rm = TRUE),
                promedio_mensual = mean(produccion_total, na.rm = TRUE),
                .groups = 'drop') %>%
      arrange(desc(produccion_total))

    tablas$consigna6_cgc_evolucion <<- cgc_cuencas_resumen_2018_2020

    graficos$p9 <<- ggplot(cgc_cuencas_evolucion_2018_2020, aes(x = año_mes, y = produccion_total, color = cuenca)) +
      geom_line(size = 1) +
      geom_point(size = 1.5) +
      labs(title = "Evolución Producción Gas por Cuenca - CGC (2018-2020)",
           x = "Fecha", y = "Producción Gas (miles m³)", color = "Cuenca") +
      theme_minimal() +
      scale_x_date(date_labels = "%Y", date_breaks = "6 months") +
      theme(axis.text.x = element_text(angle = 45, hjust = 1)) +
      scale_y_continuous(labels = comma_format())
  }

  # Consigna 7: Top 5 mediciones mensuales gas CGC (2019-2023)
  cgc_top5_mensual_gas <- gas_data_clean %>%
    filter(grepl("COMBUSTIBLES", empresa, ignore.case = TRUE),
           año >= 2019, año <= 2023) %>%
    group_by(año_mes) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  if (nrow(cgc_top5_mensual_gas) > 0) {
    tablas$consigna7_cgc_top5 <<- cgc_top5_mensual_gas

    graficos$p10 <<- ggplot(cgc_top5_mensual_gas, aes(x = reorder(format(año_mes, "%Y-%m"), produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "darkred") +
      coord_flip() +
      labs(title = "Top 5 Mediciones Mensuales Gas - CGC (2019-2023)",
           x = "Fecha", y = "Producción Gas (miles m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format())
  }

  cat("✓ Consignas de gas completadas\n")
})

# ================================
# PASO 6: ANÁLISIS COMPARATIVOS
# ================================

pipeline_step("ANÁLISIS COMPARATIVOS", {

  # Consigna 8: Top 5 empresas oil Neuquén 2024
  cat("Consigna 8: Top 5 empresas oil Neuquén 2024\n")

  top5_empresas_oil_neuquen_2024 <- oil_data_clean %>%
    filter(grepl("NEUQUÉN|NEUQUEN", provincia, ignore.case = TRUE),
           año == 2024) %>%
    group_by(empresa) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  if (nrow(top5_empresas_oil_neuquen_2024) > 0) {
    tablas$consigna8_neuquen <<- top5_empresas_oil_neuquen_2024

    graficos$p11 <<- ggplot(top5_empresas_oil_neuquen_2024, aes(x = reorder(empresa, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "gold") +
      coord_flip() +
      labs(title = "Top 5 Empresas Producción Oil - Neuquén 2024",
           x = "Empresa", y = "Producción Total Oil (m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format()) +
      theme(axis.text.y = element_text(size = 7))

    cat("✓ Consigna 8 completada\n")
  } else {
    cat("⚠ No se encontraron datos para Neuquén en 2024\n")
  }

  # Consigna 9: Top 5 empresas gas Santa Cruz 2022
  cat("Consigna 9: Top 5 empresas gas Santa Cruz 2022\n")

  top5_empresas_gas_santacruz_2022 <- gas_data_clean %>%
    filter(grepl("SANTA CRUZ", provincia, ignore.case = TRUE),
           año == 2022) %>%
    group_by(empresa) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  if (nrow(top5_empresas_gas_santacruz_2022) > 0) {
    tablas$consigna9_santacruz <<- top5_empresas_gas_santacruz_2022

    graficos$p12 <<- ggplot(top5_empresas_gas_santacruz_2022, aes(x = reorder(empresa, produccion_total), y = produccion_total)) +
      geom_bar(stat = "identity", fill = "lightgreen") +
      coord_flip() +
      labs(title = "Top 5 Empresas Producción Gas - Santa Cruz 2022",
           x = "Empresa", y = "Producción Total Gas (miles m³)") +
      theme_minimal() +
      scale_y_continuous(labels = comma_format()) +
      theme(axis.text.y = element_text(size = 7))

    cat("✓ Consigna 9 completada\n")
  } else {
    cat("⚠ No se encontraron datos para Santa Cruz en 2022\n")
  }

})

# ================================
# PASO 7: ANÁLISIS DE CONCESIONES Y DATOS FALTANTES
# ================================

pipeline_step("ANÁLISIS DE CONCESIONES Y DATOS FALTANTES", {

  # Consigna 10: Concesión con mayor producción oil 2024
  cat("Consigna 10: Concesión con mayor producción oil 2024\n")

  top_concesion_oil_2024 <- oil_data_clean %>%
    filter(año == 2024) %>%
    group_by(concesion) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(1)

  if (nrow(top_concesion_oil_2024) > 0) {
    tablas$consigna10_concesion_oil <<- top_concesion_oil_2024

    cat("✓ Concesión con mayor producción oil 2024:", top_concesion_oil_2024$concesion[1],
        "con", format(top_concesion_oil_2024$produccion_total[1], big.mark = ","), "m³\n")
  } else {
    cat("⚠ No se encontraron datos de concesiones para 2024\n")
  }

  # Consigna 11: Concesión con mayor producción gas 2022
  cat("Consigna 11: Concesión con mayor producción gas 2022\n")

  top_concesion_gas_2022 <- gas_data_clean %>%
    filter(año == 2022) %>%
    group_by(concesion) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(1)

  if (nrow(top_concesion_gas_2022) > 0) {
    tablas$consigna11_concesion_gas <<- top_concesion_gas_2022

    cat("✓ Concesión con mayor producción gas 2022:", top_concesion_gas_2022$concesion[1],
        "con", format(top_concesion_gas_2022$produccion_total[1], big.mark = ","), "miles m³\n")
  } else {
    cat("⚠ No se encontraron datos de concesiones para 2022\n")
  }

  # Consigna 12: Identificar datos faltantes
  cat("Consigna 12: Análisis de datos faltantes\n")

  # Análisis de datos faltantes en petróleo
  oil_missing <- oil_data_clean %>%
    summarise_all(~sum(is.na(.))) %>%
    gather(columna, valores_faltantes) %>%
    filter(valores_faltantes > 0)

  # Análisis de datos faltantes en gas
  gas_missing <- gas_data_clean %>%
    summarise_all(~sum(is.na(.))) %>%
    gather(columna, valores_faltantes) %>%
    filter(valores_faltantes > 0)

  # Crear resumen de datos faltantes
  missing_summary <- data.frame(
    Dataset = c("Petróleo", "Gas"),
    Total_Registros = c(nrow(oil_data_clean), nrow(gas_data_clean)),
    Registros_Completos = c(sum(complete.cases(oil_data_clean)), sum(complete.cases(gas_data_clean))),
    Registros_Con_NA = c(sum(!complete.cases(oil_data_clean)), sum(!complete.cases(gas_data_clean))),
    Porcentaje_Completos = round(c(sum(complete.cases(oil_data_clean))/nrow(oil_data_clean)*100,
                                  sum(complete.cases(gas_data_clean))/nrow(gas_data_clean)*100), 2)
  )

  tablas$consigna12_missing_summary <<- missing_summary

  if (nrow(oil_missing) > 0) {
    tablas$consigna12_oil_missing <<- oil_missing
  }

  if (nrow(gas_missing) > 0) {
    tablas$consigna12_gas_missing <<- gas_missing
  }

  # Verificar registros con producción cero o valores extraños
  oil_zeros <- sum(oil_data_clean$produccion == 0, na.rm = TRUE)
  oil_negative <- sum(oil_data_clean$produccion < 0, na.rm = TRUE)
  gas_zeros <- sum(gas_data_clean$produccion == 0, na.rm = TRUE)
  gas_negative <- sum(gas_data_clean$produccion < 0, na.rm = TRUE)

  valores_atipicos <- data.frame(
    Dataset = c("Petróleo", "Petróleo", "Gas", "Gas"),
    Tipo = c("Valores = 0", "Valores < 0", "Valores = 0", "Valores < 0"),
    Cantidad = c(oil_zeros, oil_negative, gas_zeros, gas_negative)
  )

  tablas$consigna12_valores_atipicos <<- valores_atipicos

  cat("✓ Análisis de datos faltantes completado\n")
  cat("  - Registros de petróleo completos:", sum(complete.cases(oil_data_clean)), "/", nrow(oil_data_clean), "\n")
  cat("  - Registros de gas completos:", sum(complete.cases(gas_data_clean)), "/", nrow(gas_data_clean), "\n")

})

# ================================
# PASO 8: GENERACIÓN DE ESTADÍSTICAS FINALES
# ================================

pipeline_step("GENERACIÓN DE ESTADÍSTICAS FINALES", {

  cat("Generando estadísticas generales del pipeline...\n")

  # Estadísticas generales
  stats_generales <- data.frame(
    Métrica = c("Total registros petróleo", "Total registros gas",
                "Años petróleo (min-max)", "Años gas (min-max)",
                "Empresas únicas petróleo", "Empresas únicas gas",
                "Cuencas únicas petróleo", "Cuencas únicas gas",
                "Provincias únicas petróleo", "Provincias únicas gas"),
    Valor = c(
      nrow(oil_data_clean),
      nrow(gas_data_clean),
      paste(min(oil_data_clean$año, na.rm = TRUE), "-", max(oil_data_clean$año, na.rm = TRUE)),
      paste(min(gas_data_clean$año, na.rm = TRUE), "-", max(gas_data_clean$año, na.rm = TRUE)),
      length(unique(oil_data_clean$empresa)),
      length(unique(gas_data_clean$empresa)),
      length(unique(oil_data_clean$cuenca)),
      length(unique(gas_data_clean$cuenca)),
      length(unique(oil_data_clean$provincia)),
      length(unique(gas_data_clean$provincia))
    )
  )

  tablas$estadisticas_generales <<- stats_generales

  # Top empresas productoras
  top_oil_companies <- oil_data_clean %>%
    group_by(empresa) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(10)

  top_gas_companies <- gas_data_clean %>%
    group_by(empresa) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(10)

  tablas$top_empresas_oil <<- top_oil_companies
  tablas$top_empresas_gas <<- top_gas_companies

  # Top cuencas productoras
  top_oil_basins <- oil_data_clean %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  top_gas_basins <- gas_data_clean %>%
    group_by(cuenca) %>%
    summarise(produccion_total = sum(produccion, na.rm = TRUE), .groups = 'drop') %>%
    arrange(desc(produccion_total)) %>%
    head(5)

  tablas$top_cuencas_oil <<- top_oil_basins
  tablas$top_cuencas_gas <<- top_gas_basins

  # Estadísticas de producción
  stats_produccion <- data.frame(
    Dataset = c("Petróleo (m³)", "Gas (miles m³)"),
    Media = c(mean(oil_data_clean$produccion, na.rm = TRUE),
              mean(gas_data_clean$produccion, na.rm = TRUE)),
    Mediana = c(median(oil_data_clean$produccion, na.rm = TRUE),
                median(gas_data_clean$produccion, na.rm = TRUE)),
    Máximo = c(max(oil_data_clean$produccion, na.rm = TRUE),
               max(gas_data_clean$produccion, na.rm = TRUE)),
    Mínimo = c(min(oil_data_clean$produccion, na.rm = TRUE),
               min(gas_data_clean$produccion, na.rm = TRUE)),
    Total = c(sum(oil_data_clean$produccion, na.rm = TRUE),
              sum(gas_data_clean$produccion, na.rm = TRUE))
  )

  tablas$stats_produccion <<- stats_produccion

  cat("✓ Estadísticas finales generadas\n")

})

# ================================
# PASO 9: MOSTRAR TODOS LOS RESULTADOS
# ================================

pipeline_step("MOSTRAR TODOS LOS RESULTADOS", {

  cat("Mostrando todas las tablas generadas...\n\n")

  # Mostrar todas las tablas
  for (nombre_tabla in names(tablas)) {
    cat("\n", rep("=", 50), "\n")
    cat("TABLA:", toupper(gsub("_", " ", nombre_tabla)), "\n")
    cat(rep("=", 50), "\n")

    if (is.data.frame(tablas[[nombre_tabla]])) {
      print(kable(tablas[[nombre_tabla]], digits = 2))
    } else {
      print(tablas[[nombre_tabla]])
    }
    cat("\n")
  }

  cat("\nMostrando todos los gráficos generados...\n\n")

  # Mostrar todos los gráficos
  for (nombre_grafico in names(graficos)) {
    cat("Mostrando gráfico:", nombre_grafico, "\n")
    print(graficos[[nombre_grafico]])
    cat("\n")
  }

})

# ================================
# PASO 10: FUNCIONES AUXILIARES Y EXPORTACIÓN
# ================================

pipeline_step("CONFIGURACIÓN DE FUNCIONES AUXILIARES", {

  # Función para mostrar todos los gráficos en grid
  show_all_plots_grid <<- function(plots_per_page = 4) {
    cat("Mostrando todos los gráficos en formato grid...\n")

    graficos_list <- graficos[sapply(graficos, function(x) inherits(x, "ggplot"))]

    if (length(graficos_list) > 0) {
      for (i in seq(1, length(graficos_list), by = plots_per_page)) {
        end_idx <- min(i + plots_per_page - 1, length(graficos_list))
        current_plots <- graficos_list[i:end_idx]

        if (length(current_plots) == 4) {
          grid.arrange(current_plots[[1]], current_plots[[2]],
                      current_plots[[3]], current_plots[[4]], ncol = 2, nrow = 2)
        } else if (length(current_plots) == 3) {
          grid.arrange(current_plots[[1]], current_plots[[2]],
                      current_plots[[3]], ncol = 2, nrow = 2)
        } else if (length(current_plots) == 2) {
          grid.arrange(current_plots[[1]], current_plots[[2]], ncol = 2)
        } else {
          print(current_plots[[1]])
        }
      }
    } else {
      cat("No se encontraron gráficos para mostrar.\n")
    }
  }

  # Función para exportar todos los datos procesados
  export_all_data <<- function(output_dir = "./") {
    cat("Exportando todos los datos procesados...\n")

    # Exportar datos limpios principales
    write.csv(oil_data_clean, paste0(output_dir, "datos_petroleo_procesados.csv"), row.names = FALSE)
    write.csv(gas_data_clean, paste0(output_dir, "datos_gas_procesados.csv"), row.names = FALSE)

    # Exportar todas las tablas generadas
    for (nombre_tabla in names(tablas)) {
      if (is.data.frame(tablas[[nombre_tabla]])) {
        write.csv(tablas[[nombre_tabla]],
                 paste0(output_dir, nombre_tabla, ".csv"), row.names = FALSE)
      }
    }

    cat("✓ Datos exportados exitosamente a:", output_dir, "\n")
    cat("Archivos generados:\n")
    cat("- datos_petroleo_procesados.csv\n")
    cat("- datos_gas_procesados.csv\n")
    for (nombre in names(tablas)) {
      if (is.data.frame(tablas[[nombre]])) {
        cat("-", paste0(nombre, ".csv\n"))
      }
    }
  }

  # Función para guardar todos los gráficos
  save_all_plots <<- function(output_dir = "./", width = 12, height = 8, dpi = 300) {
    cat("Guardando todos los gráficos...\n")

    for (nombre_grafico in names(graficos)) {
      if (inherits(graficos[[nombre_grafico]], "ggplot")) {
        filename <- paste0(output_dir, nombre_grafico, ".png")
        ggsave(filename, graficos[[nombre_grafico]],
               width = width, height = height, dpi = dpi)
        cat("✓ Guardado:", filename, "\n")
      }
    }

    cat("✓ Todos los gráficos guardados exitosamente\n")
  }

  # Función para generar reporte resumen
  generate_summary_report <<- function() {
    cat("\n", rep("=", 80), "\n")
    cat("REPORTE RESUMEN DEL PIPELINE DE ANÁLISIS\n")
    cat(rep("=", 80), "\n")

    cat("\n🔍 DATOS PROCESADOS:\n")
    cat("- Registros de petróleo:", nrow(oil_data_clean), "\n")
    cat("- Registros de gas:", nrow(gas_data_clean), "\n")
    cat("- Período analizado: 2018-2024\n")

    cat("\n📊 ANÁLISIS COMPLETADOS:\n")
    cat("- Tablas generadas:", length(tablas), "\n")
    cat("- Gráficos generados:", length(graficos), "\n")

    cat("\n🏢 EMPRESAS ANALIZADAS:\n")
    pampa_found <- any(grepl("PAMPA", oil_data_clean$empresa, ignore.case = TRUE))
    cgc_found <- any(grepl("COMBUSTIBLES", gas_data_clean$empresa, ignore.case = TRUE))
    cat("- PAMPA ENERGIA encontrada:", ifelse(pampa_found, "✓ SÍ", "✗ NO"), "\n")
    cat("- COMPAÑÍA GENERAL DE COMBUSTIBLES encontrada:", ifelse(cgc_found, "✓ SÍ", "✗ NO"), "\n")

    cat("\n📈 CONSIGNAS COMPLETADAS:\n")
    cat("✓ Evolución mensual PAMPA ENERGIA (2021-2023)\n")
    cat("✓ Distribución por cuencas PAMPA ENERGIA (2020-2023)\n")
    cat("✓ Distribución por yacimientos PAMPA ENERGIA (2022-2023)\n")
    cat("✓ Consigna 1: Evolución por cuencas PAMPA (2018-2023)\n")
    cat("✓ Consigna 2: Top 5 mediciones mensuales PAMPA (2018-2023)\n")
    cat("✓ Consigna 3: Evolución mensual gas CGC (2018-2023)\n")
    cat("✓ Consigna 4: Distribución por cuencas gas CGC (2021-2022)\n")
    cat("✓ Consigna 5: Distribución por yacimientos gas CGC (2019-2023)\n")
    cat("✓ Consigna 6: Evolución por cuencas gas CGC (2018-2020)\n")
    cat("✓ Consigna 7: Top 5 mediciones mensuales gas CGC (2019-2023)\n")
    cat("✓ Consigna 8: Top 5 empresas oil Neuquén 2024\n")
    cat("✓ Consigna 9: Top 5 empresas gas Santa Cruz 2022\n")
    cat("✓ Consigna 10: Concesión mayor producción oil 2024\n")
    cat("✓ Consigna 11: Concesión mayor producción gas 2022\n")
    cat("✓ Consigna 12: Análisis datos faltantes\n")

    cat("\n🛠 FUNCIONES DISPONIBLES:\n")
    cat("- show_all_plots_grid() - Mostrar gráficos en cuadrícula\n")
    cat("- export_all_data() - Exportar todos los datos\n")
    cat("- save_all_plots() - Guardar todos los gráficos\n")
    cat("- generate_summary_report() - Este reporte\n")

    cat("\n", rep("=", 80), "\n")
    cat("🎉 PIPELINE COMPLETADO EXITOSAMENTE!\n")
    cat(rep("=", 80), "\n\n")

    # Mostrar estadísticas clave
    if ("stats_produccion" %in% names(tablas)) {
      cat("📊 ESTADÍSTICAS CLAVE DE PRODUCCIÓN:\n")
      print(kable(tablas$stats_produccion, digits = 2))
      cat("\n")
    }

    if ("estadisticas_generales" %in% names(tablas)) {
      cat("📋 ESTADÍSTICAS GENERALES:\n")
      print(kable(tablas$estadisticas_generales))
      cat("\n")
    }
  }

  cat("✓ Funciones auxiliares configuradas\n")

})

# ================================
# EJECUCIÓN COMPLETA DEL PIPELINE
# ================================

cat("\n", rep("🚀", 30), "\n")
cat("INICIANDO PIPELINE COMPLETO DE ANÁLISIS\n")
cat(rep("🚀", 30), "\n\n")

# El pipeline se ejecuta automáticamente paso a paso
# Al final, mostrar el reporte resumen
generate_summary_report()

cat("\n💡 INSTRUCCIONES DE USO:\n")
cat("1. Para ver todos los gráficos: show_all_plots_grid()\n")
cat("2. Para exportar datos: export_all_data()\n")
cat("3. Para guardar gráficos: save_all_plots()\n")
cat("4. Para ver una tabla específica: print(kable(tablas$nombre_tabla))\n")
cat("5. Para ver un gráfico específico: print(graficos$nombre_grafico)\n")

cat("\n📋 TABLAS DISPONIBLES:\n")
for (nombre in names(tablas)) {
  cat("- tablas$", nombre, "\n", sep = "")
}

cat("\n📈 GRÁFICOS DISPONIBLES:\n")
for (nombre in names(graficos)) {
  cat("- graficos$", nombre, "\n", sep = "")
}

cat("\n✨ ¡ANÁLISIS COMPLETADO! ✨\n")
```
